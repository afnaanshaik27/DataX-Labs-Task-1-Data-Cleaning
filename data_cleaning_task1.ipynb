```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("marketing_campaign.csv", sep="\t")

df.head()
```

```python
print("Dataset Shape:", df.shape)

df.info()
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
median_income = df["Income"].median()

df["Income"] = df["Income"].fillna(median_income)

print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
```

```python
df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df.shape)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python
df["dt_customer"] = pd.to_datetime(
    df["dt_customer"],
    format="%d-%m-%Y",
    errors="coerce"
)

print(df["dt_customer"].dtype)
```

```python
print("Invalid Dates:", df["dt_customer"].isnull().sum())
```

```python
print("Education Values:")
print(df["education"].unique())

print("\nMarital Status Values:")
print(df["marital_status"].unique())
```

```python
text_columns = ["education", "marital_status"]

for column in text_columns:
    df[column] = df[column].str.strip().str.lower()

print("Standardized Education Values:")
print(df["education"].unique())

print("\nStandardized Marital Status Values:")
print(df["marital_status"].unique())
```

```python
print(df.dtypes)
```

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

negative_values = {}

for column in numeric_columns:
    count = (df[column] < 0).sum()
    
    if count > 0:
        negative_values[column] = count

print("Columns with Negative Values:")
print(negative_values)
```

```python
df.describe()
```

```python
print("Records with Year_Birth before 1920:")
print(df[df["year_birth"] < 1920][["id", "year_birth"]])
```

```python
df = df[df["year_birth"] >= 1920]

print("Dataset Shape After Removing Invalid Birth Years:", df.shape)
```

```python
print("Final Missing Values:")
print(df.isnull().sum())

print("\nFinal Duplicate Rows:", df.duplicated().sum())

print("\nFinal Dataset Shape:", df.shape)
```

```python
print(df.dtypes)
```

```python
df.to_csv("cleaned_marketing_campaign.csv", index=False)

print("Cleaned dataset exported successfully!")
```

## Data Cleaning Summary

The marketing campaign dataset initially contained 2,240 rows and 29 columns.

The following preprocessing steps were performed:

- Identified missing values using Pandas.
- Found 24 missing values in the Income column.
- Filled missing Income values using the median income.
- Checked for duplicate records and found no duplicate rows.
- Standardized column names by converting them to lowercase.
- Standardized categorical text values by removing extra spaces and converting text to lowercase.
- Converted the dt_customer column to datetime format.
- Checked for invalid date values and found none.
- Checked numerical columns for negative values and found none.
- Identified an unrealistic birth year and removed records with year_birth before 1920.
- Identified potential outliers, including a very high income value, but retained them because they were not confirmed as invalid.
- Exported the final cleaned dataset as cleaned_marketing_campaign.csv.

The cleaned dataset is now structured and ready for further analysis and visualization.
